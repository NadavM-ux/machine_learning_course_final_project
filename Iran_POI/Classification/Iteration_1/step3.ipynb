```python
# Step 3 — Phase A: split 100 users 50/50 between annotators
import pandas as pd
from pathlib import Path

df = pd.read_csv('../../data/Candidates_user_data_MERGED.csv')

# Sample 100 users with a fixed seed (reproducible)
sample = df.sample(n=100, random_state=42).reset_index(drop=True)
sample['profile_url'] = 'https://x.com/' + sample['username'].astype(str)

# Add empty label columns
sample['target_population']      = ''
sample['locals_vs_diaspora']     = ''
sample['person_vs_organization'] = ''
sample['comments']               = ''

# Reorder columns: useful info first, then label columns
cols = ['username', 'profile_url', 'display_name', 'description', 'location',
        'followers_count', 'following_count', 'statuses_count', 'created_at',
        'target_population', 'locals_vs_diaspora', 'person_vs_organization', 'comments']
sample = sample[cols]

# Split 50/50
nadav_half   = sample.iloc[:50].copy()   # rows 0-49
partner_half = sample.iloc[50:].copy()   # rows 50-99

# Save both files
out_dir = Path('../../Classification/Iteration_1')
out_dir.mkdir(parents=True, exist_ok=True)

nadav_half.to_csv(out_dir / 'iteration_1_labels_NADAV.csv', index=False)
partner_half.to_csv(out_dir / 'iteration_1_labels_PARTNER.csv', index=False)

print(f"Saved {len(nadav_half)} rows to iteration_1_labels_NADAV.csv")
print(f"Saved {len(partner_half)} rows to iteration_1_labels_PARTNER.csv")
print(f"\nFiles in: {out_dir.resolve()}")


nadav   = pd.read_csv('../../Classification/Iteration_1/iteration_1_labels_NADAV.csv')
partner = pd.read_csv('../../Classification/Iteration_1/iteration_1_labels_PARTNER.csv')

print(f"NADAV file:   {len(nadav)} rows, first user: {nadav['username'].iloc[0]}")
print(f"PARTNER file: {len(partner)} rows, first user: {partner['username'].iloc[0]}")
print(f"Overlap: {set(nadav['username']) & set(partner['username'])}")

import pandas as pd
test = pd.read_csv('../../Classification/Iteration_1/iteration_1_labels_NADAV.csv')
print(test.shape)
test.head()
```
