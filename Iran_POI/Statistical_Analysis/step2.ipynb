```python
import pandas as pd
df = pd.read_csv('../data/Candidates_user_data_MERGED.csv') # load the data set of the merged candidates user data, which contains the user data of all candidates, including those who have been merged.
df.head() # display the first few rows of the data frame to check the data.
```

```python
unique_users = df['username'].nunique()  # Count all unique users
print(f"Total unique users: {unique_users}") 
```

```python
followers_stats = df['followers_count'].describe()
print(followers_stats)
```

```python

df['created_at_dt'] = pd.to_datetime(df['created_at'], format='%B %Y', errors='coerce')


print("--- Stats for Following and Statuses ---")
stats_other = df[['following_count', 'statuses_count']].describe()
print(stats_other)


years = df['created_at_dt'].dt.year
print("\n--- Stats for Created At (Years) ---")
print(years.describe())
```
