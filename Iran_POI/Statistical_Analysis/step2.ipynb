```python
import pandas as pd
df = pd.read_csv('../data/Candidates_user_data_MERGED.csv') # load the data set of the merged candidates user data, which contains the user data of all candidates, including those who have been merged.
df.head() # display the first few rows of the data frame to check the data.
```

```python
unique_users = df['username'].nunique()  # Count all unique users
print(f"Total unique users: {unique_users}") 
```

```python
followers_stats = df['followers_count'].describe()
print(followers_stats)
```

```python

df['created_at_dt'] = pd.to_datetime(df['created_at'], format='%B %Y', errors='coerce')


print("--- Stats for Following and Statuses ---")
stats_counts = df[['followers_count', 'following_count', 'statuses_count']].describe()
print(stats_counts)

years = df['created_at_dt'].dt.year
print("\n--- Stats for Created At (Years) ---")
print(years.describe())
```

```python
import pandas as pd

# All the stats you already computed
followers = df['followers_count'].dropna()
following = df['following_count'].dropna()
statuses  = df['statuses_count'].dropna()
years     = df['created_at_dt'].dt.year.dropna()

summary_rows = [
    ('Total unique users',          df['username'].nunique()),
    ('', ''),  # blank row for readability
    ('statuses_count — min',        f'{statuses.min():.0f}'),
    ('statuses_count — max',        f'{statuses.max():.0f}'),
    ('statuses_count — mean',       f'{statuses.mean():.2f}'),
    ('statuses_count — median',     f'{statuses.median():.2f}'),
    ('statuses_count — std',        f'{statuses.std():.2f}'),
    ('', ''),
    ('followers_count — min',       f'{followers.min():.0f}'),
    ('followers_count — max',       f'{followers.max():.0f}'),
    ('followers_count — mean',      f'{followers.mean():.2f}'),
    ('followers_count — median',    f'{followers.median():.2f}'),
    ('followers_count — std',       f'{followers.std():.2f}'),
    ('', ''),
    ('following_count — min',       f'{following.min():.0f}'),
    ('following_count — max',       f'{following.max():.0f}'),
    ('following_count — mean',      f'{following.mean():.2f}'),
    ('following_count — median',    f'{following.median():.2f}'),
    ('following_count — std',       f'{following.std():.2f}'),
    ('', ''),
    ('created_at — earliest year',  f'{int(years.min())}'),
    ('created_at — latest year',    f'{int(years.max())}'),
    ('created_at — mean year',      f'{years.mean():.2f}'),
    ('created_at — median year',    f'{int(years.median())}'),
    ('created_at — std (years)',    f'{years.std():.2f}'),
]

summary = pd.DataFrame(summary_rows, columns=['Metric', 'Value'])
summary.to_csv('author_statistics.csv', index=False)
summary
```

```python
import matplotlib.pyplot as plt

cols = [
    ('followers_count', 'followers_histogram.png',  '#3a6ea5'),
    ('following_count', 'following_histogram.png',  '#8e44ad'),
    ('statuses_count',  'statuses_histogram.png',   '#16a085'),
]

for col, fname, color in cols:
    s = df[col].dropna()
    fig, ax = plt.subplots(figsize=(9, 5))
    ax.hist(s, bins=50, color=color, edgecolor='white')
    ax.set_title(f'Distribution of {col} (Iran POI candidates)')
    ax.set_xlabel(col)
    ax.set_ylabel('Number of users')
    fig.tight_layout()
    fig.savefig(fname, dpi=150)
    plt.show()
```
